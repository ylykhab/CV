## Задача 1

### Выборки размера 10 и 5

```python
n1 = 10 #мю=0, сигма=1
n2 = 5  #мю=1, сигма=1

x1 = np.random.normal(loc=0, scale=1, size=n1) #N(0, 1)
x2 = np.random.normal(loc=1, scale=1, size=n2) #N(1, 1)

x = np.concatenate([x1, x2])

np.random.shuffle(x)

print(x)
```

```python
mu1, mu2 = np.random.choice(x.flatten(), 2, replace=False)
var1, var2 = 1, 1
pi1, pi2 = 0.5, 0.5

def normal_pdf(x, mean, var):
    return (1 / np.sqrt(2 * np.pi * var)) * np.exp(-0.5 * ((x - mean) ** 2) / var)

tol = 1e-6
max_iter = 100
log_likelihood_old = 0

for iteration in range(max_iter):
    r1 = pi1 * normal_pdf(x, mu1, var1)
    r2 = pi2 * normal_pdf(x, mu2, var2)

    sum_r = r1 + r2
    gamma1 = r1 / sum_r
    gamma2 = r2 / sum_r

    pi1 = np.mean(gamma1)
    pi2 = np.mean(gamma2)

    mu1 = np.sum(gamma1 * x.flatten()) / np.sum(gamma1)
    mu2 = np.sum(gamma2 * x.flatten()) / np.sum(gamma2)

    log_likelihood = np.sum(np.log(sum_r))

    if np.abs(log_likelihood - log_likelihood_old) < tol:
        break
    log_likelihood_old = log_likelihood

labels = (gamma2 > gamma1).astype(int)
count1 = np.sum(labels == 0)
count2 = np.sum(labels == 1)

print(f"Матожидание компоненты 1: {mu1}, количество чисел: {count1}")
print(f"Матожидание компоненты 2: {mu2}, количество чисел: {count2}")
print("\nВероятности принадлежности к компоненте 1:")
print(gamma1)
print("\nВероятности принадлежности к компоненте 2:")
print(gamma2)
```

## Задача 2

### Генерация последовательностей

```python
states = ['a', 'b']
initial_probs = np.array([0.5, 0.5])
transition_matrix = np.array([
    [0.7, 0.3],
    [0.6, 0.4]
])

num_sequences = 1000
sequence_length = 3000

initial_states = np.random.choice(states, size=num_sequences, p=initial_probs)
sequences = np.empty((num_sequences, sequence_length), dtype=str)
sequences[:, 0] = initial_states

for i in range(1, sequence_length):
    prev_indices = (sequences[:, i-1] == 'a').astype(int)
    probs = transition_matrix[prev_indices]
    sequences[:, i] = np.where(
        np.random.rand(num_sequences) < probs[:, 0], 'a', 'b'
    )

unique, counts = np.unique(sequences[:, 0], return_counts=True)
estimated_initial_probs = counts / num_sequences

transition_counts = np.zeros((2, 2))
state_index = {'a': 0, 'b': 1}
prev_states = sequences[:, :-1].ravel()
next_states = sequences[:, 1:].ravel()
for p, n in zip(prev_states, next_states):
    transition_counts[state_index[p], state_index[n]] += 1

estimated_transition_matrix = transition_counts / transition_counts.sum(axis=1, keepdims=True)

print("Исходные вероятности начальных состояний:", initial_probs)
print("Оцененные вероятности начальных состояний:", estimated_initial_probs)

print("Исходная матрица переходов:")
print(transition_matrix)

print("Оцененная матрица переходов:")
print(estimated_transition_matrix)
```

### Сравнение моделей

```python
flattened_seq = sequences.ravel()
n = len(flattened_seq)

#бернулли
counts = Counter(flattened_seq)
total = sum(counts.values())
bernoulli_probs = {k: v / total for k, v in counts.items()}

log_likelihood_bernoulli = sum([v * np.log(bernoulli_probs[k]) for k, v in counts.items()])
k_bernoulli = len(bernoulli_probs)
AIC_bernoulli = 2 * k_bernoulli - 2 * log_likelihood_bernoulli
BIC_bernoulli = k_bernoulli * np.log(n) - 2 * log_likelihood_bernoulli

#марковская 1
transition_counts_1 = np.zeros((2, 2))
state_index = {'a': 0, 'b': 1}

for i in range(1, len(flattened_seq)):
    prev_idx = state_index[flattened_seq[i-1]]
    curr_idx = state_index[flattened_seq[i]]
    transition_counts_1[prev_idx, curr_idx] += 1

transition_probs_1 = transition_counts_1 / transition_counts_1.sum(axis=1, keepdims=True)

log_likelihood_1 = np.sum(transition_counts_1 * np.log(transition_probs_1, where=(transition_probs_1>0)))
k_1 = 2 * (2 - 1)
AIC_1 = 2 * k_1 - 2 * log_likelihood_1
BIC_1 = k_1 * np.log(n) - 2 * log_likelihood_1

#марковская 2
transition_counts_2 = defaultdict(lambda: Counter())

for i in range(2, len(flattened_seq)):
    prev_state = (flattened_seq[i-2], flattened_seq[i-1])
    curr_state = flattened_seq[i]
    transition_counts_2[prev_state][curr_state] += 1

transition_probs_2 = {
    k: {kk: vv / sum(v.values()) for kk, vv in v.items()}
    for k, v in transition_counts_2.items()
}

log_likelihood_2 = sum([
    count * np.log(transition_probs_2[k][kk])
    for k, v in transition_counts_2.items()
    for kk, count in v.items()
    if transition_probs_2[k][kk] > 0
])
k_2 = len(transition_probs_2) * (2 - 1)
AIC_2 = 2 * k_2 - 2 * log_likelihood_2
BIC_2 = k_2 * np.log(n) - 2 * log_likelihood_2

#марковская 3
transition_counts_3 = defaultdict(lambda: Counter())

for i in range(3, len(flattened_seq)):
    prev_state = (flattened_seq[i-3], flattened_seq[i-2], flattened_seq[i-1])
    curr_state = flattened_seq[i]
    transition_counts_3[prev_state][curr_state] += 1

transition_probs_3 = {
    k: {kk: vv / sum(v.values()) for kk, vv in v.items()}
    for k, v in transition_counts_3.items()
}

log_likelihood_3 = sum([
    count * np.log(transition_probs_3[k][kk])
    for k, v in transition_counts_3.items()
    for kk, count in v.items()
    if transition_probs_3[k][kk] > 0
])
k_3 = len(transition_probs_3) * (2 - 1)
AIC_3 = 2 * k_3 - 2 * log_likelihood_3
BIC_3 = k_3 * np.log(n) - 2 * log_likelihood_3


results = pd.DataFrame({
    'Model': ['Bernoulli', 'Markov 1', 'Markov 2', 'Markov 3'],
    'AIC': [AIC_bernoulli, AIC_1, AIC_2, AIC_3],
    'BIC': [BIC_bernoulli, BIC_1, BIC_2, BIC_3]
})

print(results)
```

Лучше всего описывает данные Марковская модель 3-го порядка, тк имеет наименьшие критерии AIC и BIC.

```python
def read_fasta(filename):
    with open(filename, 'r') as file:
        lines = file.readlines()
    return ''.join([line.strip() for line in lines if not line.startswith('>')])

def split_sequence(sequence, chunk_size=1000):
    return [sequence[i:i + chunk_size] for i in range(0, len(sequence), chunk_size) if len(sequence[i:i + chunk_size]) == chunk_size]

#бернулли
def bernoulli_model(sequences):
    counts = Counter(''.join(sequences))
    total = sum(counts.values())
    probs = {k: v / total for k, v in counts.items()}
    log_likelihood = sum([v * np.log(probs[k]) for k, v in counts.items()])
    k = len(probs)
    return log_likelihood, k

#марковская
def markov_model(sequences, order):
    transition_counts = defaultdict(lambda: Counter())

    for seq in sequences:
        for i in range(order, len(seq)):
            prev_state = tuple(seq[i - order:i])
            curr_state = seq[i]
            transition_counts[prev_state][curr_state] += 1

    transition_probs = {
        k: {kk: vv / sum(v.values()) for kk, vv in v.items()}
        for k, v in transition_counts.items()
    }

    log_likelihood = sum([
        count * np.log(transition_probs[k][kk])
        for k, v in transition_counts.items()
        for kk, count in v.items()
        if transition_probs[k][kk] > 0
    ])

    k = len(transition_probs) * (len(set(''.join(sequences))) - 1)
    return log_likelihood, k

def calculate_aic_bic(log_likelihood, k, n):
    AIC = 2 * k - 2 * log_likelihood
    BIC = k * np.log(n) - 2 * log_likelihood
    return AIC, BIC

def process_fasta(filename):
    sequence = read_fasta(filename)
    sequences = split_sequence(sequence, 1000)
    n = sum([len(seq) for seq in sequences])

    results = []

    log_likelihood, k = bernoulli_model(sequences)
    AIC, BIC = calculate_aic_bic(log_likelihood, k, n)
    results.append(['Bernoulli', AIC, BIC])

    for order in range(1, 6):
        log_likelihood, k = markov_model(sequences, order)
        AIC, BIC = calculate_aic_bic(log_likelihood, k, n)
        results.append([f'Markov {order}', AIC, BIC])

    return pd.DataFrame(results, columns=['Model', 'AIC', 'BIC'])

results = process_fasta('sequence.fasta')
print(results)
```

Лучше всего подошла марковская модель 5 порядка, так как имеет наименьшие значения AIC и BIC
